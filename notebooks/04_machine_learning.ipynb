Train the models, then measure what the complexity actually bought.
Note that every split is over participants. The leaky baseline at the end exists purely to quantify the error that participant-level splitting prevents.


```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
```

```python
from src import train_models as tm, complexity_analysis as ca
```

```python
df = tm.load_features()
train, test = tm.split_by_participant(df)
print(train.subject.nunique(), 'train /', test.subject.nunique(), 'held out')
```

```python
models = tm.train_and_evaluate(train, test)
models[['model','accuracy','roc_auc','f1']]
```

```python
ladder = ca.feature_ladder(train, test)
ladder[['feature_set','n_features','accuracy','roc_auc']]
```

```python
ablation = ca.electrode_ablation(train, test)
ablation[['electrode_set','n_electrodes','roc_auc','random_auc_mean','placement_advantage']]
```

```python
honest = models.set_index('model').loc['logistic_regression','roc_auc']
leaky = tm.leaky_baseline(df)['roc_auc']
print(f'participant split {honest:.3f} vs epoch split {leaky:.3f} -> inflation {leaky-honest:+.3f}')
```
